#### season quarter

````python
def season_quarter(x):
  if(x=='quarter_1'):
    return 'Jan-Mar'
  elif(x=='quarter_2'):
    return 'Apr-June'
  elif(x=='quarter_3'):
    return 'Jul-Sep'
  else:
    return 'Oct-Dec'
````

````python
dfcpy['season_quarter'] = dfcpy['season_quarter'].apply(lambda x: season_quarter(x))
````

````python
# visualize for quarter season according to hotel type
d=dfcpy.groupby(by=['hotel','season_quarter']).agg({'arrival_month':'count'}).reset_index()
sns.set_theme(style='whitegrid')
plt.figure(figsize=(8,5))
sns.barplot(x='season_quarter',y='arrival_month', hue='hotel', data=d, palette="ch:rot=-.25,hue=1,light=.75")
plt.xlabel('')
plt.ylabel('')
plt.show()
````
